# conditional with loop

```python
arr=range(5)
print(arr)
```

```python
for i in arr:
    print(i)
```

```python
for i in range(1,51):
    if i%5==0:
     print(i)
```

```python
for i in range(1,51):
    if i%2==0:
     print(i)
```

```python
for i in range(1,51):
    if i%2!=0:
     print(i)
```

```python
for i in range(1,51):
    if i==5:
     break
    print(i)
```

```python
for i in range(1,51):
    if i==5:
     print("Found")
     continue
    print("Not Found") 
```

```python
for i in range(1,51):
    if i%3==0 or i%5==0:
        continue
    print(i)
print("end")
```

# print using pattern

```python
for i in range(1,11):
    print('# # # #')
```

```python
for i in range(1,11):
    if i<=5:
     print('# # # #')
```

```python
for i in range(1,11):
    if i<=5:
     print('#')
```

```python
for i in range(1,5):
    for j in range(i):
         print('#',end='')
    print()
```

```python
for i in range(4):
    for j in range(i+1):
         print('#',end='')
    print()
```

```python
for i in range(4):
    for j in range(i+1):
         print('#',end='')
    print()
```

```python
for i in range(4):
    for j in range(4-i):
         print('#',end='')
    print()
```

```python
for i in range(1,5):
    print('#')
```

```python
for i in range(4):
    for j in range(i):
        print('#')
    print()
```

```python
for i in range(4):
     for j in range(i+1):
        print("#", end =" ")
     print()
```
